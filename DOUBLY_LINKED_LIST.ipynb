```python
class node: # Defines a node for the doubly linked list
  def __init__(self,value):
    self.value=value # Stores the data value of the node
    self.next=None # Pointer to the next node in the list
    self.prev=None # Pointer to the previous node in the list

class double_linked_list: # Defines the doubly linked list structure
  def __init__(self,value):
    new_node=node(value) # Create a new node with the given value
    self.head=new_node # Head points to the new node
    self.tail=new_node # Tail points to the new node
    self.length=1 # Initialize length to 1

  def append(self,value): # Adds a new node to the end of the list
    new_node=node(value) # Create new node
    if self.length==0: # If the list is empty
      self.tail=new_node # New node becomes both head and tail
      self.head=new_node
    else:  # If the list is not empty
      self.tail.next=new_node # Link current tail's next to new node
      new_node.prev=self.tail # Link new node's prev to current tail
      self.tail=new_node # Update tail to the new node
    self.length+=1 # Increment list length

  def print_list(self): # Prints all values in the list from head to tail
    temp=self.head # Start from the head
    while temp is not None: # Iterate until the end of the list
      print(temp.value) # Print the current node's value
      temp=temp.next # Move to the next node

  def prepend(self,value): # Adds a new node to the beginning of the list
    new_node=node(value) # Create new node
    if self.length==0: # If the list is empty
      self.head=new_node # New node becomes both head and tail
      self.tail=new_node
    else: # If the list is not empty
      self.head.prev=new_node # Link current head's prev to new node
      new_node.next=self.head # Link new node's next to current head
      self.head=new_node # Update head to the new node
    self.length+=1 # Increment list length

  def get_index(self,index): # Retrieves the node at a specific index
    if index<0 or index>=self.length: # Check for valid index
      return None # Return None if index is out of bounds
    temp=self.head # Start from the head
    for i in range (index): # Traverse to the desired index
      temp=temp.next
    return temp # Return the node at the specified index

  def insert(self,index,value): # Inserts a new node at a specific index
    if index<0 or index>self.length: # Handle invalid index
      return False # Indicate insertion failure
    if index==0: # If inserting at the beginning
      self.prepend(value) # Use prepend method
      return True
    if index==self.length: # If inserting at the end
      self.append(value) # Use append method
      return True

    new_node=node(value) # Create a new node
    temp=self.get_index(index-1) # Get the node just before the insertion point
    new_node.prev=temp # Link new node's prev to temp
    new_node.next=temp.next # Link new node's next to temp's next
    temp.next.prev=new_node # Link the node after temp to new_node's prev
    temp.next=new_node # Link temp's next to new node
    self.length+=1 # Increment list length
    return True





linked_list=double_linked_list(3) # Initialize a new double linked list
linked_list.append(4) # Append value 4
linked_list.append(5) # Append value 5
linked_list.prepend(2) # Prepend value 2
linked_list.prepend(0) # Prepend value 0
linked_list.insert(1,1) # Insert value 1 at index 1
linked_list.print_list() # Print the current list
print(f"the length of double linked list is :{linked_list.length}") # Print the length of the list
```
