#`HASH MAP`

```python
class hash_tables:
  def __init__(self,size=7):
    # Initialize the data_map (array of buckets) with a specified size.
    # Each element is initially None.
    self.data_map=[None]*size

  def __hash(self,key):
    # Private helper method to calculate the hash index for a given key.
    my_hash=0
    for letter in key:
      # Sum the ASCII values of characters, multiplied by a prime number (23) to reduce collisions.
      # Apply modulo operation with the size of data_map to ensure the index stays within bounds.
      my_hash=(my_hash+ord(letter)*23)%(len(self.data_map))
    return my_hash

  def set(self, key, value):
    # Calculates the hash index for the key.
    index = self.__hash(key)
    # If the bucket at the calculated index is empty, initialize it as an empty list.
    if self.data_map[index] == None:
      self.data_map[index] = []
    # Iterate through the list (bucket) at the current index to check if the key already exists.
    for i, (k, v) in enumerate(self.data_map[index]):
      if k == key:
        # If the key exists, update its corresponding value and return.
        self.data_map[index][i] = (key, value)
        return
    # If the key does not exist in the bucket, append the new key-value pair to the list.
    self.data_map[index].append((key, value))

  def get(self, key):
    # Calculates the hash index for the key.
    index = self.__hash(key)
    # Check if the bucket at the calculated index is not empty.
    if self.data_map[index] is not None:
      # Iterate through the key-value pairs in the bucket.
      for k, v in self.data_map[index]:
        if k == key:
          # If the key is found, return its associated value.
          return v
    # If the key is not found or the bucket is empty, return None.
    return None

  def keys(self):
    # Initialize an empty list to store all keys.
    all_keys = []
    # Iterate through each bucket in the data_map.
    for i in range(len(self.data_map)):
      # If a bucket is not empty (i.e., contains key-value pairs).
      if self.data_map[i] is not None:
        # Iterate through each key-value pair in the current bucket.
        for k, _ in self.data_map[i]:
          # Append only the key to the all_keys list.
          all_keys.append(k)
    # Return the list containing all collected keys.
    return all_keys
```

### Demonstration of `hash_tables` Class

```python
# Create a new hash table
my_hash_table = hash_tables()

# Set some key-value pairs
print("Setting key-value pairs:")
my_hash_table.set("apples", 10)
my_hash_table.set("oranges", 5)
my_hash_table.set("bananas", 20)
my_hash_table.set("grapes", 15)
my_hash_table.set("apples", 12) # Update an existing key
my_hash_table.set("lemons", 8)

# Print the internal data map to see how it's stored
print("\nInternal data map:", my_hash_table.data_map)

# Get values using the get method
print("\nGetting values:")
print(f"Apples: {my_hash_table.get('apples')}") # Should be 12
print(f"Oranges: {my_hash_table.get('oranges')}") # Should be 5
print(f"Pears: {my_hash_table.get('pears')}")   # Should be None

# Get all keys using the keys method
print("\nAll keys:", my_hash_table.keys())
```
