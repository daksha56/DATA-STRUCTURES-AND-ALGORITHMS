```python
class node:
    def __init__(self,value):
        self.value=value
        self.next=None

class linkedlist:
    def __init__(self,value):
        new_node=node(value)
        self.value=value
        self.head=new_node
        self.tail=new_node
        self.length=1
    def print_list(self):
      temp=self.head
      while temp is not None:
        print(temp.value)
        temp=temp.next
    def append(self,value):
        new_node=node(value)
        if self.head is None:
            self.head=new_node
            self.tail=new_node
        else:
            self.tail.next=new_node  #here we are assigning the tail->next=new node this is poiinting the previous node to next node
            self.tail=new_node    #here we are setiing up the tail value to appended node
        self.length+=1
    def prepend(self,value):
        new_node=node(value)
        if self.length == 0: # if the list is empty it will set head and tail to it
            self.head=new_node
            self.tail=new_node
        else:
            new_node.next=self.head
            self.head=new_node
        self.length+=1
    def get(self,index):
        if index<0 or index>=self.length:
            return None
        temp=self.head
        for _ in range(index):
            temp=temp.next
        return temp
    def set_value(self,index,value):
        temp=self.get(index)
        if temp is not None:
            temp.value=value
            return True
        return False
    def insert(self,index,value) :
        if index<0 or index>self.length:
            return False
        if index==0:
            return self.prepend(value)
        if index==self.length:
            return self.append(value)
        new_node=node(value)
        temp=self.get(index-1)
        new_node.next=temp.next
        temp.next=new_node
        self.length += 1 # Increment length after insertion
        return True # Return True for successful insertion
    def pop(self):
      if self.length == 0:
        return None
      if self.length == 1:
        popped_value = self.head.value
        self.head = None
        self.tail = None
        self.length = 0
        return popped_value

      temp = self.head
      pre = self.head
      while temp.next is not None:
        pre = temp
        temp = temp.next

      popped_value = temp.value
      self.tail = pre
      self.tail.next = None
      self.length -= 1
      return popped_value
    def pop_first(self):
      if self.length == 0:
        return None
      temp = self.head
      if self.length == 1:
        self.head = None
        self.tail = None
      else:
        self.head = self.head.next
        temp.next = None
      self.length -= 1
      return temp.value

    def remove(self,index):
      if index<0 or index>self.length:
        return None
      if index==0:
        return self.pop_first()
      if index==self.length-1 :
        return self.pop()
      pre=self.get(index-1)
      temp=pre.next
      removed_value = temp.value
      pre.next=temp.next
      temp.next=None
      self.length-=1
      return removed_value

    def reverse(self):
      # Swap head and tail initially
      temp = self.head
      self.head = self.tail
      self.tail = temp

      # Initialize pointers for reversal
      after = temp.next
      before = None

      # Iterate through the list, reversing pointers
      for _ in range(self.length):
        after = temp.next  # Save the next node
        temp.next = before # Reverse the current node's pointer
        before = temp      # Move 'before' one step forward
        temp = after       # Move 'temp' one step forward



linked_list=linkedlist(1)
linked_list.append(2)
linked_list.append(3)
linked_list.prepend(0)
linked_list.get(2)
linked_list.set_value(2,6)
linked_list.insert(2,4)

print('Original List:')
linked_list.print_list()

linked_list.reverse()

print('\nReversed List:')
linked_list.print_list()

# Original popping demonstration (kept for completeness if needed later)
print('List before popping:')
linked_list.print_list()
print('Popped value:', linked_list.pop())
print('Popped value:', linked_list.pop_first())
print('List after popping:')
linked_list.print_list()
```
