```python
class node:
    def __init__(self,value):
        self.value=value
        self.next=None

class linkedlist:
    def __init__(self,value):
        new_node=node(value)
        self.value=value
        self.head=new_node
        self.tail=new_node
        self.length=1
    def print_list(self):
      temp=self.head
      while temp is not None:
        print(temp.value)
        temp=temp.next
    def append(self,value):
        new_node=node(value)
        if self.head is None:
            self.head=new_node
            self.tail=new_node
        else:
            self.tail.next=new_node  #here we are assigning the tail->next=new node this is poiinting the previous node to next node
            self.tail=new_node    #here we are setiing up the tail value to appended node
        self.length+=1
    def prepend(self,value):
        new_node=node(value)
        if self.length == 0: # if the list is empty it will set head and tail to it
            self.head=new_node
            self.tail=new_node
        else:
            new_node.next=self.head
            self.head=new_node
        self.length+=1
    def get(self,index):
        if index<0 or index>self.length:
            return None
        temp=self.head
        for i in range(index):
            temp=temp.next
        return temp
    def set_value(self,index,value):
        temp=self.get(index)
        if temp is not None:
            temp.value=value
            return True
        return False
    def insert(self,index,value) :
        if index<0 or index>self.length: # Changed self.index to index
            return False
        if index==0: # Changed self.index to index
            return self.prepend(value)
        if index==self.length: # Changed self.index to index
            return self.append(value)
        new_node=node(value)
        temp=self.get(index-1)
        new_node.next=temp.next
        temp.next=new_node







linked_list=linkedlist(1)
linked_list.append(2)
linked_list.append(3)
linked_list.prepend(0)
linked_list.get(2)
linked_list.set_value(2,6)
linked_list.insert(2,4)
linked_list.print_list()
```
