```python
class node:
    def __init__(self,value):
        # Constructor for the node class
        # Each node stores a 'value' and a reference to the 'next' node in the list
        self.value=value
        self.next=None

class linkedlist:
    def __init__(self,value):
        # Constructor for the linkedlist class
        # Initializes a new linked list with a single node
        new_node=node(value)
        self.value=value # This 'value' attribute is redundant, as the head node already holds it
        self.head=new_node # The first node in the list
        self.tail=new_node # The last node in the list
        self.length=1 # Keeps track of the number of nodes in the list
    def print_list(self):
      # Prints the value of each node in the linked list from head to tail
      temp=self.head
      while temp is not None:
        print(temp.value)
        temp=temp.next
    def append(self,value):
        # Adds a new node with the given 'value' to the end of the list
        new_node=node(value)
        if self.head is None: # If the list is empty, the new node becomes both head and tail
            self.head=new_node
            self.tail=new_node
        else:
            self.tail.next=new_node  # The current tail's 'next' pointer points to the new node
            self.tail=new_node    # The new node becomes the new tail
        self.length+=1 # Increment the length of the list
    def prepend(self,value):
        # Adds a new node with the given 'value' to the beginning of the list
        new_node=node(value)
        if self.length == 0: # If the list is empty, the new node becomes both head and tail
            self.head=new_node
            self.tail=new_node
        else:
            new_node.next=self.head # The new node's 'next' pointer points to the current head
            self.head=new_node # The new node becomes the new head
        self.length+=1 # Increment the length of the list
    def get(self,index):
        # Retrieves the node at the specified 'index'
        # Returns None if the index is out of bounds
        if index<0 or index>=self.length:
            return None
        temp=self.head
        for _ in range(index):
            temp=temp.next # Traverse the list until the desired index is reached
        return temp
    def set_value(self,index,value):
        # Sets the 'value' of the node at the specified 'index'
        # Returns True if successful, False otherwise
        temp=self.get(index)
        if temp is not None:
            temp.value=value # Update the value of the found node
            return True
        return False
    def insert(self,index,value) :
        # Inserts a new node with 'value' at the specified 'index'
        # Handles insertion at the beginning (prepend) or end (append) as special cases
        # Returns True if successful, False if index is out of bounds
        if index<0 or index>self.length:
            return False
        if index==0:
            return self.prepend(value)
        if index==self.length:
            return self.append(value)
        new_node=node(value)
        temp=self.get(index-1) # Get the node just before the insertion point
        new_node.next=temp.next # New node points to what the previous node was pointing to
        temp.next=new_node # Previous node now points to the new node
        self.length += 1 # Increment length after insertion
        return True # Return True for successful insertion
    def pop(self):
      # Removes and returns the value of the last node in the list
      # Returns None if the list is empty
      if self.length == 0:
        return None
      if self.length == 1:
        popped_value = self.head.value
        self.head = None
        self.tail = None
        self.length = 0
        return popped_value

      temp = self.head
      pre = self.head
      while temp.next is not None:
        pre = temp # 'pre' tracks the node before 'temp'
        temp = temp.next # 'temp' moves to the next node

      popped_value = temp.value # Store the value of the node to be removed
      self.tail = pre # The second to last node becomes the new tail
      self.tail.next = None # The new tail's 'next' pointer is set to None
      self.length -= 1 # Decrement the length of the list
      return popped_value
    def pop_first(self):
      # Removes and returns the value of the first node in the list
      # Returns None if the list is empty
      if self.length == 0:
        return None
      temp = self.head
      if self.length == 1:
        self.head = None
        self.tail = None
      else:
        self.head = self.head.next # The second node becomes the new head
        temp.next = None # Disconnect the original head from the list
      self.length -= 1 # Decrement the length of the list
      return temp.value # Return the value of the removed node

    def remove(self,index):
      # Removes and returns the value of the node at the specified 'index'
      # Handles removal from beginning (pop_first) or end (pop) as special cases
      # Returns None if the index is out of bounds
      if index<0 or index>self.length:
        return None
      if index==0:
        return self.pop_first()
      if index==self.length-1 :
        return self.pop()
      pre=self.get(index-1) # Get the node just before the node to be removed
      temp=pre.next # The node to be removed
      removed_value = temp.value # Store the value of the node to be removed
      pre.next=temp.next # The previous node's 'next' pointer bypasses the removed node
      temp.next=None # Disconnect the removed node from the list
      self.length-=1 # Decrement the length of the list
      return removed_value # Return the value of the removed node

    def reverse(self):
      # Reverses the order of the linked list in-place
      # Swaps the head and tail, then iterates to reverse pointers
      temp = self.head
      self.head = self.tail
      self.tail = temp

      # Initialize pointers for reversal
      after = temp.next # Will temporarily store the next node
      before = None # Will store the previously processed node

      # Iterate through the list, reversing pointers
      for _ in range(self.length):
        after = temp.next  # Save the next node before modifying current node's 'next'
        temp.next = before # Reverse the current node's pointer to point to the 'before' node
        before = temp      # Move 'before' one step forward to the current node
        temp = after       # Move 'temp' one step forward to the next node



linked_list=linkedlist(1)
linked_list.append(2)
linked_list.append(3)
linked_list.prepend(0)
linked_list.get(2)
linked_list.set_value(2,6)
linked_list.insert(2,4)

print('Original List:')
linked_list.print_list()

linked_list.reverse()

print('\nReversed List:')
linked_list.print_list()

# Original popping demonstration (kept for completeness if needed later)
print('List before popping:')
linked_list.print_list()
print('Popped value:', linked_list.pop())
print('Popped value:', linked_list.pop_first())
print('List after popping:')
linked_list.print_list()
```
