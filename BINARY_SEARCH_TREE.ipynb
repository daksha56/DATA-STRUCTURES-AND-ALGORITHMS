#BINARY SEARCH TREE

```python
class node:
  def __init__(self,value):
    self.value=value
    self.left=None
    self.right=None


class bst:
  def __init__(self):
    self.root=None
  def insert(self,value):
    new_node=node(value)
    if self.root is None:
      self.root=new_node
      return True
    temp=self.root

    while True:
      if value==temp.value:
          return False

      if value>temp.value:
        if temp.right is None:
          temp.right=new_node
          return True
      if value<temp.value:
        if temp.left is None:
          temp.left=new_node
          return True
  def search(self,value):
    if self.root is None:
      return False
    temp=self.root
    while temp is not None:
      if value==temp.value:
        return True
      elif value>temp.value:
        temp=temp.right
      elif value<temp.value:
        temp=temp.left
    return False





```

### Test Cases for BST

```python
my_bst = bst()

# Test insert method
print("--- Testing Insert ---")
print(f"Insert 10: {my_bst.insert(10)}")  # Should be True (root)
print(f"Insert 5: {my_bst.insert(5)}")   # Should be True
print(f"Insert 15: {my_bst.insert(15)}") # Should be True
print(f"Insert 3: {my_bst.insert(3)}")   # Should be True
print(f"Insert 7: {my_bst.insert(7)}")   # Should be True
print(f"Insert 12: {my_bst.insert(12)}") # Should be True
print(f"Insert 18: {my_bst.insert(18)}") # Should be True
print(f"Insert 10 again: {my_bst.insert(10)}") # Should be False (duplicate)
print(f"Insert 5 again: {my_bst.insert(5)}")   # Should be False (duplicate)

print("Root value:", my_bst.root.value)
print("Left child of root:", my_bst.root.left.value)
print("Right child of root:", my_bst.root.right.value)
```

```python
# Test search method
print("\n--- Testing Search ---")
print(f"Search for 10: {my_bst.search(10)}") # Should be True
print(f"Search for 5: {my_bst.search(5)}")   # Should be True
print(f"Search for 15: {my_bst.search(15)}") # Should be True
print(f"Search for 3: {my_bst.search(3)}")   # Should be True
print(f"Search for 7: {my_bst.search(7)}")   # Should be True
print(f"Search for 12: {my_bst.search(12)}") # Should be True
print(f"Search for 18: {my_bst.search(18)}") # Should be True

print(f"Search for 1: {my_bst.search(1)}")   # Should be False
print(f"Search for 100: {my_bst.search(100)}") # Should be False
print(f"Search for 0: {my_bst.search(0)}")   # Should be False
```
