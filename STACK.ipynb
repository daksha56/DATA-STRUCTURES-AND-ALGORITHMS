#STACK

```python
# CONSTRUCTOR
class node:
  def __init__(self,value):
    self.value=value
    self.next=None
class stack:
  def __init__(self,value):
    new_node=node(value)
    self.top=new_node
    self.height=1
#PUSH
  def push(self,value):
    new_node=node(value)
    if self.height==0:
      self.top=new_node
    new_node.next=self.top
    self.top=new_node
    self.height+=1
#POP
  def pop(self):
    if self.height==0:
      return f"Empty Stack"
    temp=self.top
    self.top=self.top.next
    temp.next=None
    self.height-=1
    return temp

#PRINT STACK FUNCTION
  def print_stack(self):
    if self.height==0:
      return f"empty stack"
    else:
      temp=self.top
      while temp is not None:
        print(temp.value)
        temp=temp.next
```

### Demonstrating Stack Functionality

```python
# Create a new stack instance
my_stack = stack(10)
print("Initial Stack:")
print(f"Top value: {my_stack.top.value}")
print(f"Height: {my_stack.height}")
```

```python
# Push elements onto the stack
my_stack.push(20)
my_stack.push(30)
my_stack.push(40)
print("\nAfter pushing 20, 30, 40:")
```

```python
# Pop remaining elements until the stack is empty
print("\nPopping all remaining elements:")
while my_stack.height > 0:
    item = my_stack.pop()
    print(f"Popped: {item.value}")

print("\nStack after popping all elements:")
print(f"Stack height: {my_stack.height}")
```
