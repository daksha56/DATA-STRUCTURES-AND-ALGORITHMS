```python
class graph:
  def __init__(self):
    self.adj_list={}
  def add_vertex(self,vertex):
    # Check if the vertex is not already in the adjacency list
    if vertex not in self.adj_list.keys():
      # If not present, add the vertex with an empty list as its neighbors
      self.adj_list[vertex]=[]
      return True
    return False
  def add_edge(self,v1,v2):
    # Check if both vertices exist in the graph
    if v1 in self.adj_list.keys() and v2 in self.adj_list.keys():
      # Add an edge from v1 to v2
      self.adj_list[v1].append(v2)
      # Add an edge from v2 to v1 (for an undirected graph)
      self.adj_list[v2].append(v1)
      return True
    return False
  def remove_edge(self,v1,v2):
    # Check if both vertices exist in the graph
    if v1 in self.adj_list.keys() and v2 in self.adj_list.keys():
      try:
        # Try to remove v2 from v1's neighbors
        self.adj_list[v1].remove(v2)
        # Try to remove v1 from v2's neighbors
        self.adj_list[v2].remove(v1)
      except ValueError:
        # Handle case where edge might not exist (e.g., if already removed or not present)
        pass

      return True
    return False
  def remove_vertex(self,vertex):
    # Check if the vertex exists in the graph
    if vertex in self.adj_list.keys():
      # Iterate through all neighbors of the vertex to be removed
      for edges in self.adj_list[vertex]:
        # Remove the edge connecting the neighbor to the target vertex
        self.adj_list[vertex].remove(edges)
      # Delete the vertex from the adjacency list
      del self.adj_list[vertex]
      return True
    return False

g = graph()
print("Initial graph:", g.adj_list)

g.add_vertex('A')
g.add_vertex('B')
g.add_vertex('C')
print("After adding vertices A, B, C:", g.adj_list)

g.add_edge('A', 'B')
g.add_edge('B', 'C')
print("After adding edges A-B, B-C:", g.adj_list)

g.remove_edge('A', 'B')
print("After removing edge A-B:", g.adj_list)

g.remove_vertex('C')
print("After removing vertex C:", g.adj_list)
```
