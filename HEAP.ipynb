```python
class Max_heap:
  def __init__(self):
    # Initialize an empty list to store the elements of the max-heap.
    self.Max_heap = []

  def left_child(self, index):
    # Calculate the index of the left child for a given parent index.
    # In a 0-indexed array, the left child of an element at index 'i' is at '2*i + 1'.
    return 2 * index + 1

  def right_child(self, index):
    # Calculate the index of the right child for a given parent index.
    # In a 0-indexed array, the right child of an element at index 'i' is at '2*i + 2'.
    return 2 * index + 2

  def _parent(self, index):
    # Calculate the index of the parent for a given child index.
    # For a 0-indexed child at 'i', the parent is at '(i - 1) // 2'.
    return (index - 1) // 2

  def swap(self, index1, index2):
    # Swap two elements in the heap array given their indices.
    # This is a common operation in heap manipulations.
    self.Max_heap[index1], self.Max_heap[index2] = self.Max_heap[index2], self.Max_heap[index1]

  def insert(self, value):
    # Add the new value to the end of the heap array.
    self.Max_heap.append(value)
    # Get the index of the newly inserted element.
    current = len(self.Max_heap) - 1

    # While the current element is not the root (index > 0)
    # AND the current element is greater than its parent (violating max-heap property):
    while current > 0 and self.Max_heap[current] > self.Max_heap[self._parent(current)]:
      # Swap the current element with its parent.
      self.swap(current, self._parent(current))
      # Move up to the parent's position to continue checking the heap property upwards.
      current = self._parent(current)

  def sink_down(self, index):
    # Assume the element at the given index is the largest initially.
    largest = index
    # Get the indices of the left and right children.
    left = self.left_child(index)
    right = self.right_child(index)
    # Get the current size of the heap.
    heap_size = len(self.Max_heap)

    # If the left child exists (within bounds) AND is greater than the current largest element:
    if left < heap_size and self.Max_heap[left] > self.Max_heap[largest]:
      # Update 'largest' to the index of the left child.
      largest = left

    # If the right child exists (within bounds) AND is greater than the current largest element:
    if right < heap_size and self.Max_heap[right] > self.Max_heap[largest]:
      # Update 'largest' to the index of the right child.
      largest = right

    # If the largest element is not the one we started with (meaning a child was larger):
    if largest != index:
      # Swap the element at 'index' with the largest child.
      self.swap(index, largest)
      # Recursively call sink_down on the new position of the element to maintain the heap property downwards.
      self.sink_down(largest)

  def remove(self, index):
    # If the heap is empty, there's nothing to remove, return None.
    if not self.Max_heap:
      return None
    # Check if the provided index is valid (within the bounds of the heap array).
    if index < 0 or index >= len(self.Max_heap):
      # If the index is out of bounds, raise an IndexError.
      raise IndexError("Index out of bounds")

    # Store the value of the element being removed.
    removed_value = self.Max_heap[index]
    # Remove the last element from the heap and store it.
    last_element = self.Max_heap.pop()

    # If the heap is not empty after popping the last element
    # AND the removed element was not the last element itself:
    if index < len(self.Max_heap): # This condition implicitly checks if the heap became empty or if 'index' was already the last element.
      # Replace the element at the 'index' where the removal occurred with the 'last_element'.
      self.Max_heap[index] = last_element
      # Call sink_down on this position to restore the max-heap property.
      self.sink_down(index)
    # Return the value that was originally removed.
    return removed_value
```

### Testing the Max_heap class

```python
my_heap = Max_heap()

# Insert elements
print("Inserting elements: 10, 5, 20, 15, 25")
my_heap.insert(10)
print(f"Heap after inserting 10: {my_heap.Max_heap}")
my_heap.insert(5)
print(f"Heap after inserting 5: {my_heap.Max_heap}")
my_heap.insert(20)
print(f"Heap after inserting 20: {my_heap.Max_heap}")
my_heap.insert(15)
print(f"Heap after inserting 15: {my_heap.Max_heap}")
my_heap.insert(25)
print(f"Heap after inserting 25: {my_heap.Max_heap}")

print("\nHeap after all insertions:")
display(my_heap.Max_heap)
```

```python
# Remove an element (e.g., the root, which is at index 0)
# In a max heap, removing from index 0 usually means extracting max, then re-heapifying.
# The current remove function removes an arbitrary index. Let's test removing the max (index 0).

if my_heap.Max_heap:
  print(f"\nRemoving element at index 0 (current max: {my_heap.Max_heap[0]})\n")
  removed = my_heap.remove(0)
  print(f"Removed value: {removed}")
  print(f"Heap after removing element at index 0: {my_heap.Max_heap}")

if len(my_heap.Max_heap) > 2:
  print(f"\nRemoving element at index 1 (value: {my_heap.Max_heap[1]})\n")
  removed = my_heap.remove(1)
  print(f"Removed value: {removed}")
  print(f"Heap after removing element at index 1: {my_heap.Max_heap}")
else:
  print("\nHeap has too few elements to test removing at index 1.")

print("\nFinal heap state:")
display(my_heap.Max_heap)
```
