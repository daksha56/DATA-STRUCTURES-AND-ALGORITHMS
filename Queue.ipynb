#QUEUE

```python
class node:
  def __init__(self,value):
    self.value=value
    self.next=None

class queue:
  def __init__(self,value=None):
    if value is None:
      self.head=None
      self.length=0
    else:
      new_node=node(value)
      self.head=new_node
      self.length=1

  def enq(self,value):
    new_node=node(value)
    if self.head is None:
      self.head=new_node
    else:
      temp=self.head
      while temp.next is not None:
        temp=temp.next
      temp.next=new_node
    self.length+=1

  def deque(self):
    if self.length==0:
      return None

    removed_value = self.head.value
    if self.length==1:
      self.head=None
    else:
      self.head=self.head.next
    self.length-=1
    return removed_value

  def print_q(self):
    elements = []
    temp = self.head
    while temp is not None:
      elements.append(temp.value)
      temp = temp.next

    # Print in reverse order (last in, first out for printing)
    print("Queue elements (last to first):")
    for element in reversed(elements):
      print(element)


# --- Test Cases for Corrected Queue ---
print("\n--- Testing q1 (initialized with value) ---")
q1=queue(1)
print("Enqueuing 2, 3, 4:")
q1.enq(2)
q1.enq(3)
q1.enq(4)
q1.print_q() # Expected: 4, 3, 2, 1

print("\nDequeuing an element:", q1.deque()) # Expected: 1
print("Queue elements after first dequeue (last to first):")
q1.print_q() # Expected: 4, 3, 2

print("\nDequeuing another element:", q1.deque()) # Expected: 2
print("Queue elements after second dequeue (last to first):")
q1.print_q() # Expected: 4, 3

print("\n--- Testing q2 (initialized empty) ---")
q2 = queue() # Test empty initialization
print("Enqueuing 10, 20, 30:")
q2.enq(10)
q2.enq(20)
q2.enq(30)
print("Queue 2 elements (last to first):")
q2.print_q() # Expected: 30, 20, 10

print("\nDequeuing from empty queue test:")
q_empty = queue()
print("Attempting to dequeue from empty:", q_empty.deque()) # Expected: None
q_empty.enq(5)
print("Enqueued 5, then dequeued:", q_empty.deque()) # Expected: 5
print("Attempting to dequeue from now empty:", q_empty.deque()) # Expected: None
```
