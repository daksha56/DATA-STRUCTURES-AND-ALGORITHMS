#RECURSION

```python
# CALL BACK
def three():
  print("three")
def two():
  three()
  print("two")
def one():
  two()
  print("one")


# one calling one it directs it to function two ,two directs to function three so function three gets executed first than 2 than one
one()
```

```python
#FACTORIAL
def factorial(num):
  if num==0:
    return 1
  else:
    return num*(factorial(num-1)) # its calling num*factorial(num-1) until num is 0
                                  # let num=3  3*factorial(3-1)
                                  # 3*factorial(2)
                                  #3*2*factorial(1)
                                  #3*2*1*factorial(0)
                                  #factorial(0) return 1
                                  # then call back happens and 3*2*1*1 gets executed

print(factorial(6))
print(factorial(5))
print(factorial(4))
print(factorial(3))
print(factorial(1))
print(factorial(0))
```
